```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Mall_Customers.csv") #Dataset file name
dataset #Function call
```

```python
x=dataset.iloc[:,[3,4]].values #Separate the column and rows
```

```python
from sklearn.cluster import Birch
X = [[0, 1], [0.3, 1], [-0.3, 1], [0, -1], [0.3, -1], [-0.3, -1]]
brc = Birch(n_clusters=None)
brc.fit(X)
brc.predict(X)
```

```python
#Model creation
from sklearn.cluster import Birch #These first two lines are called as a procedure
birch= Birch()
y_birch=birch.fit_predict(x)  #Fit - Called Function
```

```python
y_birch #Function call
```

```python
supervised=pd.DataFrame(dataset)
supervised['Cluster_group']=y_birch
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
centroids=birch
centroids #Function Call
```

```python
import seaborn as sns #Seaborn is a library - used for graphical purposes
facet=sns.lmplot(data=supervised, x=supervised.columns[3],y=supervised.columns[4],hue=supervised.columns[5],
                 fit_reg=False,legend=True,legend_out=True)
```
