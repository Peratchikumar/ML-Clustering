```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Mall_Customers.csv") #Dataset file name
dataset #Function call
```

```python
x=dataset.iloc[:,[3,4]].values #Separate the column and rows
```

```python
from sklearn.cluster import MeanShift
X = np.array([[1, 1], [2, 1], [1, 0],
              [4, 7], [3, 5], [3, 6]])
clustering = MeanShift(bandwidth=2).fit(X)
clustering.labels_
clustering.predict([[0, 0], [5, 5]])
clustering
```

```python
#Model creation
from sklearn.cluster import MeanShift #These first two lines are called as a procedure
meanshift=MeanShift()
y_meanshift=meanshift.fit_predict(x)  #Fit - Called Function
```

```python
y_meanshift #Function call
```

```python
supervised=pd.DataFrame(dataset)
supervised['Cluster_group']=y_meanshift
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
centroids=meanshift.cluster_centers_
centroids #Function Call
```

```python
import seaborn as sns #Seaborn is a library - used for graphical purposes
facet=sns.lmplot(data=supervised, x=supervised.columns[3],y=supervised.columns[4],hue=supervised.columns[5],
                 fit_reg=False,legend=True,legend_out=True)
```
