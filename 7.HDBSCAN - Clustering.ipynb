```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Mall_Customers.csv") #Dataset file name
dataset #Function call
```

```python
x=dataset.iloc[:,[3,4]].values #Separate the column and rows
```

```python
from sklearn.cluster import HDBSCAN
from sklearn.datasets import load_digits
X, _ = load_digits(return_X_y=True)
hdb = HDBSCAN(copy=True, min_cluster_size=20)
hdb.fit(X)
hdb.labels_.shape == (X.shape[0],)
np.unique(hdb.labels_).tolist()
```

```python
#Model creation
from sklearn.cluster import HDBSCAN #These first two lines are called as a procedure
hdbscan=HDBSCAN()
y_hdbscan=hdbscan.fit_predict(x)  #Fit - Called Function
```

```python
y_hdbscan #Function call
```

```python
supervised=pd.DataFrame(dataset)
supervised['Cluster_group']=y_hdbscan
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
centroids=hdbscan
centroids #Function Call
```

```python
import seaborn as sns #Seaborn is a library - used for graphical purposes
facet=sns.lmplot(data=supervised, x=supervised.columns[3],y=supervised.columns[4],hue=supervised.columns[5],
                 fit_reg=False,legend=True,legend_out=True)
```
