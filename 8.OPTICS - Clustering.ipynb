```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Mall_Customers.csv") #Dataset file name
dataset #Function call
```

```python
x=dataset.iloc[:,[3,4]].values #Separate the column and rows
```

```python
from sklearn.cluster import OPTICS
X = np.array([[1, 2], [2, 5], [3, 6],
              [8, 7], [8, 8], [7, 3]])
clustering = OPTICS(min_samples=2).fit(X)
clustering.labels_
```

```python
#Model creation
from sklearn.cluster import OPTICS #These first two lines are called as a procedure
optics= OPTICS()
y_optics=optics.fit_predict(x)  #Fit - Called Function
```

```python
y_optics #Function call
```

```python
supervised=pd.DataFrame(dataset)
supervised['Cluster_group']=y_optics
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
centroids=optics
centroids #Function Call
```

```python
import seaborn as sns #Seaborn is a library - used for graphical purposes
facet=sns.lmplot(data=supervised, x=supervised.columns[3],y=supervised.columns[4],hue=supervised.columns[5],
                 fit_reg=False,legend=True,legend_out=True)
```
