```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Mall_Customers.csv") #Dataset file name
dataset #Function call
```

```python
x=dataset.iloc[:,[3,4]].values #Separate the column and rows
```

```python
from sklearn.cluster import SpectralClustering
clustering = SpectralClustering(n_clusters=2,
        assign_labels='discretize',
        random_state=0).fit(x)
clustering.labels_
clustering
```

```python
#Model creation
from sklearn.cluster import SpectralClustering #These first two lines are called as a procedure
spectralclustering=SpectralClustering(random_state=5)
y_spectralclustering=spectralclustering.fit_predict(x)  #Fit - Called Function
```

```python
y_spectralclustering #Function call
```

```python
supervised=pd.DataFrame(dataset)
supervised['Cluster_group']=y_spectralclustering
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
centroids=SpectralClustering
centroids #Function Call
```
